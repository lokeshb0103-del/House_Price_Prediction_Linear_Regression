```python
from re import X
import pandas as pd     ## Data manipulation
import numpy as np      ## numerical calcualations


##importing csv files

df=pd.read_csv('data.csv')
# df.head()
# df.tail()
# df.shape
# df.info()

df.isnull().sum()       ##checking missing values

df.duplicated().sum()   ##checking duplicate values

df.dtypes               ##checking data types

df.describe()

df['price'].describe()

(df['price']==0).sum()

df=df[df['price']>0]
df['price'].describe()

 ## data visualization

# import matplotlib.pyplot as plt
# plt.scatter(df['sqft_living'], df['price'])
# plt.xlabel('Living Area')
# plt.ylabel('Price')
# plt.title('Living Area vs Price')
# plt.show()

##separate X and Y

# We drop the target 'price' and non-numeric columns (like date, street, city, statezip, country)
# select_dtypes(include=[np.number]) will automatically keep only the numeric features
x=df.drop('price',axis=1).select_dtypes(include=[np.number])
y=df['price']


##Train-test split

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)
# print(x_train.shape)
# print(x_test.shape)
# print(y_train.shape)
# print(y_test.shape)

##Train the model

from sklearn.linear_model import LinearRegression

model=LinearRegression()

model.fit(x_train,y_train)

##predicting the model

y_pred=model.predict(x_test)

# print(y_pred[:5])
# print(y_test.head())

##model evaluation  using metrics r2 square,mean_absoulute_error,mean_squared_error.

from sklearn.metrics import r2_score

r2=r2_score(y_test,y_pred)
print(r2)

from sklearn.metrics import mean_absolute_error as absolute_error

mae=absolute_error(y_test,y_pred)
print(mae)

from sklearn.metrics import mean_squared_error

mse=mean_squared_error(y_test,y_pred)
print(mse)

# plt.scatter(y_test, y_pred)

# plt.xlabel("Actual Price")
# plt.ylabel("Predicted Price")
# plt.title("Actual vs Predicted House Prices")

# plt.show()
```
